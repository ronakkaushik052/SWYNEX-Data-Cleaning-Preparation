```python
import pandas as pd

url = "https://raw.githubusercontent.com/plotly/datasets/master/supermarket_Sales.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully!")
print(df.head())
```

```python
df.describe()
```

```python
print("First 5 Records:")
display(df.head())

print("Last 5 Records:")
display(df.tail())
```

```python
# Check missing values in each column

print("Missing Values:")
print(df.isnull().sum())
```

```python
total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Values Percentage:")
print(missing_percentage)
```

```python
import matplotlib.pyplot as plt

df.isnull().sum().plot(kind='bar', figsize=(12,5))

plt.title("Missing Values in Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.show()
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
duplicate_rows = df[df.duplicated()]

display(duplicate_rows)
```

```python
print("Unique Values in Each Column:")

for column in df.columns:
    print(column, ":", df[column].nunique())
```

```python
print("Column Data Types:")
print(df.dtypes)
```

```python
print("Dataset Columns:")
for column in df.columns:
    print(column)
```

```python
print("Categorical Columns:")

categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    print(column, ":", df[column].nunique())
```

```python
print("Numerical Columns:")

numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

for column in numerical_columns:
    print(column, ":", df[column].nunique())
```

```python
print("Date Column Sample:")
print(df["Date"].head(10))

print("\nTime Column Sample:")
print(df["Time"].head(5))
```

```python
print("Original Date Format:")
print(df["Date"].head(10))
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y", errors="coerce")
```

```python
print("Updated Date Data Type:", df["Date"].dtype)
print("\nDate Sample After Conversion:")
print(df["Date"].head(10))

print("\nInvalid Dates:", df["Date"].isnull().sum())
```

```python
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    print("\nColumn:", column)
    print(df[column].unique()[:10])
```

```python
text_columns = df.select_dtypes(include=["object", "str"]).columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("Extra spaces cleaned successfully!")
```

```python
for column in text_columns:
    print(column, ":", df[column].unique()[:10])
```

```python
print("Quantity Minimum:", df["Quantity"].min())
print("Quantity Maximum:", df["Quantity"].max())

print("\nUnit Price Minimum:", df["Unit price"].min())
print("Unit Price Maximum:", df["Unit price"].max())

print("\nTotal Minimum:", df["Total"].min())
print("Total Maximum:", df["Total"].max())
```

```python
print("Invalid Quantity:", (df["Quantity"] <= 0).sum())
print("Invalid Unit Price:", (df["Unit price"] <= 0).sum())
print("Invalid Total:", (df["Total"] <= 0).sum())

print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]

    print(column, ":", len(outliers), "outliers")
```

```python
import matplotlib.pyplot as plt

columns_to_plot = ["Total", "Tax 5%", "gross income"]

plt.figure(figsize=(12, 6))

df[columns_to_plot].boxplot()

plt.title("Outlier Detection using Boxplot")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
print(df.columns.tolist())
```

```python
columns_to_plot = [
    col for col in df.columns
    if col.strip().lower() in ["total", "tax 5%", "gross income"]
]

print("Selected Columns:", columns_to_plot)

df[columns_to_plot].plot(
    kind="box",
    figsize=(12, 6),
    title="Outlier Detection using Boxplot"
)

plt.ylabel("Values")
plt.grid(True)
plt.show()
```

```python
import pandas as pd

report_data = {
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Total Missing Values",
        "Duplicate Rows",
        "Invalid Dates",
        "Invalid Quantity",
        "Invalid Unit Price",
        "Invalid Total"
    ],
    "Result": [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["Date"].isnull().sum(),
        (df["Quantity"] <= 0).sum() if "Quantity" in df.columns else "N/A",
        (df["Unit price"] <= 0).sum() if "Unit price" in df.columns else "N/A",
        (df["Total"] <= 0).sum() if "Total" in df.columns else "N/A"
    ]
}

quality_report = pd.DataFrame(report_data)

display(quality_report)
```
